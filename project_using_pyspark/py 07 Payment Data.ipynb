```python
payment_schema = 'payment_id int,customer_id int,payment_date TIMESTAMP,payment_status int,payment_method STRING'
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, TimestampType
py_payment_schema = StructType([
    StructField('payment_id', IntegerType()),
    StructField('customer_id', IntegerType()),
    StructField('payment_date', TimestampType()),
    StructField('payment_status', IntegerType()),
    StructField('payment_method', StringType())
])
```

```python
df = spark.read.format("csv").option('delimiter',',').schema(py_payment_schema).load('s3://databricks-gizmobox-dl-project/external_data/payments/')
display(df)
```

```python
df.writeTo('gizmobox.bronze.py_payments').createOrReplace()
```
