```python
df_customer = spark.table('gizmobox.silver.py_customers')
df_addresses = spark.table('gizmobox.silver.py_addresses')
display(df_customer)
display(df_addresses)
```

```python
df_joinned = df_customer.join(df_addresses, df_customer.customer_id == df_addresses.customer_id, "inner").select(df_customer.customer_id, df_customer.customer_name, df_customer.date_of_birth, df_customer.email, df_customer.telephone, df_addresses.billing_address_line_1, df_addresses.billing_city, df_addresses.billing_state, df_addresses.billing_postcode, df_addresses.shipping_address_line_1, df_addresses.shipping_city, df_addresses.shipping_state, df_addresses.shipping_postcode)
display(df_joinned)
```

```python
df_joinned.writeTo('gizmobox.gold.py_customer_addresses').createOrReplace()
```
