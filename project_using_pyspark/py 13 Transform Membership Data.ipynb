```python
df = spark.table('gizmobox.bronze.py_memberships')
display(df)
```

```python
from pyspark.sql import functions as F
df_extracted = df.select(F.regexp_extract(df.path,'.*/([0-9]+)\\.png$', 1).alias('customer_id'), df.content.alias('membership_card'))
display(df_extracted)
```

```python
df_extracted.writeTo('gizmobox.silver.py_memberships').createOrReplace()
```
