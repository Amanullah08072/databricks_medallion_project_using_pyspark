```python
df = spark.table('gizmobox.bronze.py_refunds')
display(df)
```

```python
from pyspark.sql import functions as F
df_refunds_transformed = df.select('refund_id','payment_id',F.date_format('refund_timestamp','yyyy-MM-dd').cast('date').alias('refund_date'),
                                   F.date_format('refund_timestamp','HH-mm-ss').alias('refund_time'),
                                   'refund_amount',F.split(df.refund_reason,':')[0].alias('refund_reason'),
                                   F.split(df.refund_reason,':')[1].alias('refund_source'))

display(df_refunds_transformed)
```

#using regex

```python
from pyspark.sql import functions as F
df_refunds_transformed = df.select('refund_id','payment_id',F.date_format('refund_timestamp','yyyy-MM-dd').cast('date').alias('refund_date'),
                                   F.date_format('refund_timestamp','HH-mm-ss').alias('refund_time'),
                                   'refund_amount',F.regexp_extract(df.refund_reason,'^([^:]+):',1).alias('refund_reason'),
                                   F.regexp_extract(df.refund_reason,'^[^:]+:(.*)$',1).alias('refund_source'))

display(df_refunds_transformed)
```

```python
df_refunds_transformed.writeTo('gizmobox.silver.py_refunds').createOrReplace()
```
